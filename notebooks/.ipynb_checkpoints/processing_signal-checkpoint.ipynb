Plot Standard Analysis
======================

```python
std_trigger = analysis.SigmaPeakTrigger(
    sigma=3,
    required_samples_below_hysteresis=20,
    required_samples_above_threshold=20,
)

plot_kwargs = dict(
    index=5,
    n_rows=3_000_000,
    # n_rows=100_000,
    inset_ylim=0.6,
    ylim=0.6,
    figsize=(18, 6)
)

if plot_kwargs["index"] >= 4:
    plot_kwargs["inset_ylim"] = 0.6
else:
    plot_kwargs["inset_ylim"] = 0.4


match plot_kwargs["index"]:
    case 7:
        plot_kwargs["hist_lim"] = 670
    case 6:
        plot_kwargs["hist_lim"] = 670
    case 5:
        plot_kwargs["hist_lim"] = 670
    case 4:
        plot_kwargs["hist_lim"] = 670
    case 3:
        plot_kwargs["hist_lim"] = 210
    case 2:
        plot_kwargs["hist_lim"] = 61
    case 1:
        plot_kwargs["hist_lim"] = 40
    case _:
        plot_kwargs["hist_lim"] = None


start = 0.0006
delta = 0.00025 / 5
# delta = 0.00025 / 6
end = start + delta


filename = f"./data/run_5/{list_of_files[plot_kwargs["index"]][0]}"
data = analysis.CsvTrace(filename=filename, n_rows=plot_kwargs["n_rows"])

standard_analyzer = analysis.StandardTraceAnalyzer(
    std_trigger=std_trigger,
    sequence_length=wavenet.sequence_length,
    signal_normalization="zscore",
)

signal = utils.segment_signal(
    np.asarray(data.y_processed, dtype=float).ravel(),
    window_size=standard_analyzer.config.sequence_length,
)

record = standard_analyzer.analyze_processed_signal(
    signal,
    dx=data.dx,
    filename=filename,
)


bins = np.linspace(0, 1, 300)

with plt.style.context(scientific):
    _, expected_particle_flow = yo.get_record_with_expected_particle_flow(
        index=plot_kwargs["index"]
    )

    figure = record.plot_standard_detection_with_histogram(
        x_axis="time",
        figsize=plot_kwargs["figsize"],
        bins=bins,
        title=" ",
        histogram_title="Histogram",
        signal_color="C0",
        histogram_color="C0",
        histogram_line_width=0.3,
        ylim=(0, plot_kwargs["ylim"]),
        threshold_line_width=2.0,
        show_threshold=True,
        show_peaks=True,
        show_legend=False,
        expected_particle_flow=expected_particle_flow,
        show_throughput=True,
        show_inset=True,
        inset_xlim=(start, end),
        inset_ylim=(0.02, plot_kwargs["inset_ylim"]),
        histogram_reference_lines=[0.15],
        histogram_xlim=(0, plot_kwargs["hist_lim"]),
        subplots_adjust={
            "left": 0.08,
            "right": 0.98,
            "bottom": 0.14,
            "top": 0.88,
            "wspace": 0.08,
        },
    )

    figure.axes[1].grid(visible=True, which='both', axis='x', alpha=0.5)
    plt.savefig("./figures/trace_plus_hist.png", dpi=300)

    plt.show()
```

Plot CNN Analysis
=================

```python
plot_kwargs = dict(
    index=7,
    # n_rows=10_000_000,
    n_rows=3_000_000,
    # n_rows=500_000,
    inset_ylim=0.6,
    ylim=0.6,
    figsize=(18, 6)
)

match plot_kwargs["index"]:
    case 7:
        plot_kwargs["hist_lim"] = 670
    case 6:
        plot_kwargs["hist_lim"] = 670
    case 5:
        plot_kwargs["hist_lim"] = 670
    case 4:
        plot_kwargs["hist_lim"] = 670
    case 3:
        plot_kwargs["hist_lim"] = 210
    case 2:
        plot_kwargs["hist_lim"] = 61
    case 1:
        plot_kwargs["hist_lim"] = 40


match plot_kwargs["index"]:
    case 7:
        cnn_amplitude_sigma_samples = 31
    case 6:
        cnn_amplitude_sigma_samples = 31
    case 5:
        cnn_amplitude_sigma_samples = 24
    case 4:
        cnn_amplitude_sigma_samples = 24
    case 3:
        cnn_amplitude_sigma_samples = 30
    case 2:
        cnn_amplitude_sigma_samples = 30
    case 1:
        cnn_amplitude_sigma_samples = 30
    case _:
        cnn_amplitude_sigma_samples = 34


if plot_kwargs["index"] >= 4:
    plot_kwargs["inset_ylim"] = 0.6
else:
    plot_kwargs["inset_ylim"] = 0.4


cnn_trigger = analysis.HeightPeakTrigger(
    height=0.50,
    required_samples_below_hysteresis=4,
    required_samples_above_threshold=2,
)

filename = f"./data/run_5/{list_of_files[plot_kwargs["index"]][0]}"
data = analysis.CsvTrace(
    filename=filename,
    n_rows=plot_kwargs["n_rows"]
)

cnn_analyzer = analysis.CNNTraceAnalyzer(
    wavenet=wavenet,
    cnn_trigger=cnn_trigger,
    signal_normalization="zscore",
    cnn_amplitude_baseline=0.06,
    cnn_amplitude_sigma_samples=cnn_amplitude_sigma_samples,
)

signal = utils.segment_signal(
    np.asarray(data.y_processed, dtype=float).ravel(),
    window_size=cnn_analyzer.config.sequence_length,
)

record = cnn_analyzer.analyze_processed_signal(
    signal,
    dx=data.dx,
    filename=filename,
)

bins = np.linspace(0, 1, 300)





expected_particle_flow = yo.get_expected_particle_flow(index=plot_kwargs["index"])
with plt.style.context(scientific):
    _, expected_particle_flow = yo.get_record_with_expected_particle_flow(
        index=plot_kwargs["index"]
    )

    figure = record.plot_wavenet_detection_with_histogram(
        x_axis="time",
        figsize=plot_kwargs["figsize"],
        bins=bins,
        title=" ",
        histogram_title="Histogram",
        prediction_color="C0",
        signal_color="C1",
        histogram_color="C1",
        histogram_line_width=0.3,
        ylim=(0, plot_kwargs["ylim"]),
        threshold_line_width=2.0,
        show_cnn_prediction_peaks=False,
        show_cnn_signal_peaks=False,
        show_legend=False,
        show_cnn_recovered_signal_peaks=True,

        show_prediction=False,
        show_cnn_threshold=False,
        show_cnn_reconstructed_trace=False,

        expected_particle_flow=expected_particle_flow,
        show_throughput=True,
        show_inset=True,
        inset_xlim=(start, end),
        inset_ylim=(0.02, plot_kwargs["inset_ylim"]),
        histogram_reference_lines=[0.15],
        histogram_xlim=(0, plot_kwargs["hist_lim"]),
        subplots_adjust={
            "left": 0.08,
            "right": 0.98,
            "bottom": 0.14,
            "top": 0.88,
            "wspace": 0.08,
        },
        show=False
    )

    figure.axes[1].set_ylim([0, 0.6])
    figure.axes[1].grid(visible=True, which='both', axis='x', alpha=0.5)
    figure.savefig("figures/trace_plus_hist.png", dpi=300)

    plt.show()
```

Dilution Series
===============

```python
dilution_series = analysis.DilutionSeries(
    folder="./data/run_5",
    wavenet=wavenet,
    initial_concentration=1e9,
    nrows=500_000,
    std_trigger=std_trigger,
    cnn_amplitude_sigma_samples=27,
    cnn_trigger=cnn_trigger,
    water_filename="_water_1.csv",
    trace_files=list_of_files[:8]
)

result = dilution_series.run()
```

Time of arrival study
=====================

```python
with plt.style.context(scientific):
    index = 6
    expected_particle_flow = dilution_series.get_expected_particle_flow(index=index)

    figure, axes = plt.subplots(ncols=1, nrows=2, figsize=(9, 11), sharex=True)

    throughput_label = (
        f"{expected_particle_flow / 1000:.1f}k" if expected_particle_flow >= 1000 and expected_particle_flow < 1_000_000
        else f"{expected_particle_flow / 1_000_000:.1f}M" if expected_particle_flow >= 1_000_000
        else f"{expected_particle_flow:.0f}"
    )
    figure.suptitle(f"Throughput: {throughput_label} / second", y=1.0, x=0.35)

    figure.supylabel("", fontsize=30)


    bins = np.linspace(0, 0.0012, 15300)

    dilution_series.poisson.plot.histogram(
        index=index,
        # detector="cnn",
        ax=axes[0],
        detector="standard",
        bins=bins,
        title="",
        edge_color="black",
        xlim_quantile=0.99,
        histogram_color="C0",
        edge_line_width=1.0,
        x_axis="time",
        show=False,
        figsize=(12, 6)
    )

    dilution_series.poisson.plot.histogram(
        index=index,
        detector="cnn",
        ax=axes[1],
        bins=bins,
        title="",
        xlim_quantile=0.99,
        edge_color="black",
        histogram_color="C1",
        edge_line_width=1.0,
        x_axis="time",
        show=False,
        figsize=(12, 6)
    )

    axes[1].set_ylabel("")
    axes[1].set_title("WaveNet", loc='right')

    axes[0].set_title("Standard", loc='right')
    axes[0].set_ylabel("")



    factor = 1000
    for ax in figure.axes:
        ax.set_xlabel("")

        ax.xaxis.set_major_formatter(
            FuncFormatter(lambda value, position: f"{value * factor:g}")
        )

    # axes[0].set_xlim([None, 0.02 * 10 / factor])
    axes[-1].set_xlabel("Inter-arrival time [ms]", fontsize=30)
    plt.subplots_adjust(hspace=1)
    plt.tight_layout()

    plt.savefig("figures/time_of_arrival.png", bbox_inches='tight')

    plt.show()
```

Amplitude retrieval study
=========================

```python
with plt.style.context(scientific):
    index = 5
    expected_particle_flow = dilution_series.get_expected_particle_flow(index=index)

    figure, axes = plt.subplots(ncols=1, nrows=2, figsize=(9, 11), sharex=True)

    throughput_label = (
        f"{expected_particle_flow / 1000:.1f}k" if expected_particle_flow >= 1000 and expected_particle_flow < 1_000_000
        else f"{expected_particle_flow / 1_000_000:.1f}M" if expected_particle_flow >= 1_000_000
        else f"{expected_particle_flow:.0f}"
    )
    figure.suptitle(f"Throughput: {throughput_label} / second", y=1.0, x=0.35)

    figure.supylabel("", fontsize=30)
    bins = np.linspace(0, 1, 300)


    dilution_series.amplitude.plot.histogram(
        index=index,
        ax=axes[0],
        detector="standard",
        # bins=200,
        bins=bins,
        title="",
        xlim_quantile=0.99,
        edge_color="black",
        histogram_color="C0",
        edge_line_width=1.0,
        show=False,
        figsize=(12, 6)
    )

    dilution_series.amplitude.plot.histogram(
        index=index,
        detector="cnn",
        ax=axes[1],
        # bins=200,
        bins=bins,
        title="",
        edge_color="black",
        histogram_color="C1",
        edge_line_width=1.0,
        show=False,
        figsize=(12, 6)
    )


    axes[1].set_ylabel("")
    axes[1].set_title("WaveNet", loc='right')

    axes[0].set_title("Standard", loc='right')
    axes[0].set_ylabel("")


    factor = 1000
    for ax in figure.axes:
        ax.set_xlabel("")

        ax.xaxis.set_major_formatter(
            FuncFormatter(lambda value, position: f"{value * factor:g}")
        )



    plt.tight_layout()
    plt.savefig("figures/amplitudes.png")

    plt.show()
```
