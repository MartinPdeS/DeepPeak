# DeepPeak maintained tutorial

This tutorial generates reproducible traces, evaluates the classical baseline, and shows the handoff to WaveNet. It tracks the current public API.

```python
from DeepPeak.benchmarking import classical_detector, evaluate_detector, make_benchmark_dataset

cases = make_benchmark_dataset(traces_per_level=4, seed=42)
results = evaluate_detector("Classical", classical_detector(), cases)
[result.to_dict() for result in results[:3]]
```
